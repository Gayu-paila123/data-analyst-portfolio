```python
import pandas as pd

data = {
    "Order_ID": [101, 102, 103, 104, 105],
    "Product": [
        "Laptop", "Mouse", "Keyboard",
        "Monitor", "Headphones"
    ],
    "Quantity": [2, 5, 3, 2, 4],
    "Price": [45000, 500, 1200, 12000, 2000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]

print(df)
```

```python
df.head()
```

```python
total_revenue = df["Revenue"].sum()

print("Total Revenue:", total_revenue)
```

```python
product_sales = df.groupby("Product")["Quantity"].sum()

print(product_sales)
```

```python
top_products = df.sort_values(
    by="Revenue",
    ascending=False
)

print(top_products[["Product", "Revenue"]])
```

```python
df.sort_values(by="Revenue", ascending=True)
```

```python
average_revenue = df["Revenue"].mean()

print("Average Revenue:", average_revenue)
```

```python
print("RETAIL SALES REPORT")
print("-------------------")

print("Total Revenue:", df["Revenue"].sum())
print("Total Units Sold:", df["Quantity"].sum())
print("Average Revenue:", df["Revenue"].mean())
print("Number of Products:", df["Product"].nunique())
```

```python
import matplotlib.pyplot as plt

plt.bar(df["Product"], df["Revenue"])

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (₹)")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(df["Product"], df["Quantity"])

plt.title("Units Sold by Product")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")

plt.xticks(rotation=45)
plt.show()
```
