```python
df.info()
```

```python
total_revenue = df["Revenue"].sum()

print("Total Revenue:", total_revenue)
```

```python
total_units = df["Quantity"].sum()

print("Total Units Sold:", total_units)
```

```python
average_order_value = df["Revenue"].mean()

print("Average Order Value:", average_order_value)
```

```python
product_sales = df.groupby("Product")["Revenue"].sum()

product_sales = product_sales.sort_values(ascending=False)

print(product_sales)
```

```python
import matplotlib.pyplot as plt

product_sales.plot(kind="bar", figsize=(9,5))

plt.title("Total Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("product_revenue_analysis.png")
plt.show()
```

```python
region_sales = df.groupby("Region")["Revenue"].sum()

region_sales = region_sales.sort_values(ascending=False)

print(region_sales)
```

```python
region_sales.plot(kind="bar", figsize=(8,5))

plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("regional_revenue_analysis.png")
plt.show()
```

```python
category_sales = df.groupby("Category")["Revenue"].sum()

print(category_sales)
```

```python
category_sales.plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("category_revenue_analysis.png")
plt.show()
```
